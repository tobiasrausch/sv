# Browsing the tutorial SVs

Please run the cells top to bottom.

```python
import os
import igv_notebook
DATA = "." if os.path.exists("genome.fa") else "data/lr"
igv_notebook.init()
```

```python
b = igv_notebook.Browser({
    "reference": {
        "id": "HG008-chr1-chr5",
        "fastaURL": f"{DATA}/genome.fa",
        "indexURL": f"{DATA}/genome.fa.fai",
    },
    "locus": "chr5:56,631,000-56,633,000",
})

# Structural variant landmarks
b.load_track({
    "name": "SVs",
    "url": f"{DATA}/svs.bed",
    "format": "bed",
    "type": "annotation",
    "displayMode": "EXPANDED",
})

# Tumor and control alignments (CRAM, decoded against the reference above),
# coloured & grouped by haplotype (HP tag)
for sample in ("tumor", "control"):
    b.load_track({
        "name": f"{sample} (HP)",
        "url": f"{DATA}/{sample}.cram",
        "indexURL": f"{DATA}/{sample}.cram.crai",
        "format": "cram",
        "type": "alignment",
        "colorBy": "tag",
        "colorByTag": "HP",
        "groupBy": "tag",
        "groupByTag": "HP",
        "height": 300,
    })
```
