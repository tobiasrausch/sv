# Browsing the tutorial SVs

Please run the cells top to bottom.

```python
import os
import igv_notebook
DATA = "." if os.path.exists("genome.fa") else "data/lr"
igv_notebook.init()
```

```python
b = igv_notebook.Browser({
    "reference": {
        "id": "Genome",
        "fastaURL": f"{DATA}/genome.fa",
        "indexURL": f"{DATA}/genome.fa.fai",
    },
    "locus": "chr5:12,807,462-12,824,809",
})

# Structural variants
b.load_track({
    "name": "SVs",
    "url": f"{DATA}/svs.bed",
    "format": "bed",
    "type": "annotation",
    "displayMode": "EXPANDED",
})

# Tumor and control alignments (change 'none' to 'tag' for HP-tagging)
for sample in ("tumor", "control"):
    b.load_track({
        "name": f"{sample}",
        "url": f"{DATA}/{sample}.cram",
        "indexURL": f"{DATA}/{sample}.cram.crai",
        "format": "cram",
        "type": "alignment",
        "colorBy": "none",
        "colorByTag": "HP",
        "height": 300,
    })
```
