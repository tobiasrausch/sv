# Browsing the tutorial SVs

This notebook embeds [igv.js](https://github.com/igvteam/igv.js) to browse the modBAMs in the browser.

Please run the cells top to bottom.

```python
import igv_notebook

DATA = "data/lr"
igv_notebook.init()
```

```python
b = igv_notebook.Browser({
    "reference": {
        "id": "Tutorial",
        "fastaURL": f"{DATA}/genome.fa",
        "indexURL": f"{DATA}/genome.fa.fai",
    },
    "locus": "chr5:56,631,000-56,633,000",
})

# Structural variants
b.load_track({
    "name": "SVs",
    "url": f"{DATA}/svs.bed",
    "format": "bed",
    "type": "annotation",
    "displayMode": "EXPANDED",
})

# Tumor and control alignments
for sample in ("tumor", "control"):
    b.load_track({
        "name": f"{sample} (HP)",
        "url": f"{DATA}/{sample}.hp.bam",
        "indexURL": f"{DATA}/{sample}.hp.bam.bai",
        "format": "bam",
        "type": "alignment",
        "colorBy": "tag",
        "colorByTag": "HP",
        "groupBy": "tag",
        "groupByTag": "HP",
        "height": 300,
    })
```
